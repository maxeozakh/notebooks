```python
import torch

# define inputs
x = torch.tensor([2.0, 3.0], requires_grad=True)
f1 = x[0] + x[1]
f2 = x[0] * x[1]
f = torch.stack([f1, f2])


J = []
for i in range(2):
    grads = torch.autograd.grad(f[i], x, retain_graph=True)
    J.append(grads[0])
J = torch.stack(J)
'''
J = [
    [d(f1)/d(x), d(f1)/d(y)],
    [d(f2)/d(x), d(f2)/d(y)]
]
'''
print("Jacobian:\n", J)
```

```python
delta = torch.tensor([
     0.1, # x change, Δx
    -0.2  # y change, Δy
]) 


delta_f = J @ delta

'''
d(f1)/d(x) * Δx + d(f1)/d(y) * Δy <-- total change of f1, after wigling knobs
d(f2)/d(x) * Δx + d(f2)/d(y) * Δy <-- total change of f2


1 * .1 + 1 * -.2   -->   .1 - .2   -->    -0.1
3 * .1 + 2 * -.2         .3 - .4          -0.1 
'''
print(delta_f)
```

* Jacobian matrixes are regular one, with specific content
* Specifically, they contain *partial* derivatives of functions with respect to their inputs
* We can multiply this Jacobian matrix by a map of the changes for each input, and get an answer for "how much each function output will change, with respect to changes to it's each argument"

It works only for small steps! Here is Why analogy:
* Imagine standing on a curved hill and looking at the slope right where you stand
* That slope is the Jacobian value
* If you take a tiny step, it tells you pretty well how your height will change
* But if you take 100 steps without re-arangement, the hill might bend, go down, or up again — and your straight-line estimate will be wrong

Summary:
* Jacobian gives a linear approximation at a point
* Exact if the function is linear
* Works well for small changes
* Fails for large changes on nonlinear functions
